```python
from jarvis.db.figshare import data
import pandas as pd
import matplotlib.pyplot as plt
```

```python
mxene_data = data("mxene275")
df = pd.DataFrame(mxene_data)


print(df.columns)
df.head()
```

```python
# Check missing values
print(df.isnull().sum())
```

```python
plt.hist(df["formation_energy"], bins=30, color="steelblue", edgecolor="black")
plt.xlabel("Formation energy (eV)")
plt.ylabel("Count")
plt.title("Distribusi Formation Energy MXene")
plt.show()
```
